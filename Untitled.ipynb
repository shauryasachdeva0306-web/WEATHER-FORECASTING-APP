# INPUT

```python
inputs=Input(shape=(sequence_length,1))

x=LSTM(64,return_sequences=True)(inputs)
x=Dropout(0.2)(x)
x=LSTM(64)(x)
x=Dropout(0.2)(x)
```

# OUTPUT

```python
temp_output=Dense(1,name='temperature_output')(x)
condition_output=Dense(len(label.classes_),activation='softmax',name='condition_output')(x)
```

```python
model=Model(inputs=inputs,outputs=[temp_output,condition_output])
```

```python
model.compile(
    optimizer='adam',loss={'temperature_output':['mae'],
                           'condition_output':['accuracy']})
```
